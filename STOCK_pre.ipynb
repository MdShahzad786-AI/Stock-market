## Feature Engineering

I will create the following new features:

1.  **Time-based features**: Day of the week, month, and year extracted from the 'date' column.
2.  **Daily Returns**: The percentage change in the 'close' price.
3.  **Price Volatility/Range**: The difference between 'high' and 'low' prices.
4.  **Average Price**: The average of 'open', 'high', 'low', and 'close' prices.
5.  **Moving Averages**: 7-day and 30-day simple moving averages of the 'close' price. These will be calculated per ticker.

```python
# Extract time-based features
df['day_of_week'] = df['date'].dt.dayofweek
df['month'] = df['date'].dt.month
df['year'] = df['date'].dt.year

# Calculate daily returns (percentage change in close price)
# Group by 'ticker' to calculate returns correctly for each stock
df['daily_return'] = df.groupby('ticker')['close'].pct_change() * 100

# Calculate price range (high - low)
df['price_range'] = df['high'] - df['low']

# Calculate average price
df['avg_price'] = (df['open'] + df['high'] + df['low'] + df['close']) / 4

# Calculate moving averages (7-day and 30-day SMA)
# Group by 'ticker' before applying rolling window to avoid mixing stocks
df['SMA_7'] = df.groupby('ticker')['close'].transform(lambda x: x.rolling(window=7).mean())
df['SMA_30'] = df.groupby('ticker')['close'].transform(lambda x: x.rolling(window=30).mean())

# Display the DataFrame with new features
print('DataFrame with new engineered features:')
display(df.head())
```

## Model Training: Predicting Future Closing Price (Regression)

We will use a LightGBM Regressor model to predict the future closing price of the stocks. This involves several steps:

1.  **Define Target Variable**: Create a `future_close` column, which is the 'close' price shifted one day forward for each ticker.
2.  **Handle Missing Values**: Drop rows that now have `NaN`s due to the shifting and rolling mean calculations.
3.  **One-Hot Encode Categorical Features**: Convert the 'exchange' column into numerical format.
4.  **Feature Selection**: Choose the features that will be used to train the model.
5.  **Time-Series Train-Test Split**: Split the data chronologically to ensure our model is evaluated on future, unseen data.
6.  **Train LightGBM Regressor**: Train the model.
7.  **Evaluate Model**: Assess the model's performance using metrics like Mean Absolute Error (MAE) and R-squared.
8.  **Check Model Size**: Verify the model size against the 20MB constraint.

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, r2_score
import lightgbm as lgb
import joblib
import os
import pickle

# 1. Define Target Variable: Predict next day's closing price
df['future_close'] = df.groupby('ticker')['close'].shift(-1)

# Drop rows where 'future_close' is NaN (last day for each ticker)
df.dropna(subset=['future_close'], inplace=True)

# Drop other NaNs that resulted from feature engineering (e.g., initial SMA values)
df.dropna(inplace=True)

print('DataFrame after creating target and handling NaNs:')
display(df.head())
```

```python
# 2. Feature and Target Selection

# Features to use for training
features = [
    'open', 'high', 'low', 'close', 'volume',
    'day_of_week', 'month', 'year',
    'daily_return', 'price_range', 'avg_price',
    'SMA_7', 'SMA_30', 'exchange'
]

X = df[features]
y = df['future_close']

# Identify categorical and numerical features for preprocessing
categorical_features = ['exchange']
numerical_features = [f for f in features if f not in categorical_features]

# Create a column transformer for one-hot encoding categorical features
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ],
    remainder='passthrough' # Keep numerical features as they are
)

print("Features and target defined. Preprocessor created.")
```

```python
# 3. Time-Series Train-Test Split
# We'll use data up to a certain date for training and after that for testing.
# Using 2025-01-01 as a split point for demonstration.

split_date = pd.to_datetime('2025-01-01')

X_train = X[df['date'] < split_date]
y_train = y[df['date'] < split_date]
X_test = X[df['date'] >= split_date]
y_test = y[df['date'] >= split_date]

print(f"Training set size: {X_train.shape[0]} samples")
print(f"Testing set size: {X_test.shape[0]} samples")
```

```python
# 4. Train LightGBM Regressor Model

# Create a pipeline that first preprocesses features and then trains the model
model_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', lgb.LGBMRegressor(random_state=42, n_estimators=1000, learning_rate=0.05, num_leaves=31, n_jobs=-1))
    # Hyperparameters are examples, tuning might be required for optimal performance
])

print("Training LightGBM model...")
model_pipeline.fit(X_train, y_train)
print("Model training complete.")
```

```python
# 5. Evaluate Model Performance

y_pred = model_pipeline.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"R-squared (R2): {r2:.4f}")
```

```python
# 6. Check Model Size

model_filename = 'lgbm_stock_predictor.pkl'
with open(model_filename, 'wb') as file:
    pickle.dump(model_pipeline, file)

model_size_bytes = os.path.getsize(model_filename)
model_size_mb = model_size_bytes / (1024 * 1024)

print(f"Model saved as: {model_filename}")
print(f"Model size: {model_size_mb:.2f} MB")

if model_size_mb < 20:
    print("Model size is within the 20 MB limit.")
else:
    print("WARNING: Model size exceeds the 20 MB limit. Consider tuning hyperparameters or using a simpler model.")
```
